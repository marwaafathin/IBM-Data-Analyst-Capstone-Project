#### Step 1: Import the necessary module.


### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
## Summarize dataset column data types and counts
df.info()

## Missing value info
print(df.isna().sum())
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
df.describe()
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Identify inconsistent entries in Country
print(df['Country'].nunique())
print(sorted(df['Country'].dropna().unique()))
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Standarized country

country_map = {
    'United State of America': 'United States',
    'United Kingdom of Great Britain and Northern Ireland': 'United Kingdom',
    'Russian Federation': 'Russia',
    'Republic of Korea': 'South Korea'
}

df['Country'] = df['Country'].replace(country_map)

print(df['Country'].isin(country_map.keys()).sum())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## One-Hot encoding Employment
employment_encoded = pd.get_dummies(df['Employment'], prefix='Employment', dtype=int)

df = pd.concat([df, employment_encoded], axis=1)
df[employment_encoded.columns].head()
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Identify columns with the highest number of missing values

missing = df.isna().sum().sort_values(ascending=False)
print(missing.head(10))
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Impute missing values in ConvertedCompYearly with median
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(df['ConvertedCompYearly'].median())

print("Missing ConvertedCompYearly after:", df['ConvertedCompYearly'].isna().sum())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Impute missing values in RemoteWork with the most frequent value
df['RemoteWork'] = df['RemoteWork'].fillna(df['RemoteWork'].mode()[0])

print("Missing RemoteWork after:", df['RemoteWork'].isna().sum())
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Apply Min-Max Scalling to ConvertedCompYearly
col = df['ConvertedCompYearly']
df['ConvertedCompYearly_MinMax'] = (col - col.min()) / (col.max() - col.min())

df['ConvertedCompYearly_MinMax'].describe()
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Log-transform ConvertedCompYearly to reduce skewness
import numpy as np

df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])

print("Skewness before:", df['ConvertedCompYearly'].skew())
print("Skewness after:", df['ConvertedCompYearly_Log'].skew())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Create ExperienceLevel based on YearsCodePro
df['YearsCodePro'] = df['YearsCodePro'].replace({
    'Less than 1 year': 0,
    'More than 50 years': 50
})
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

df['ExperienceLevel'] = pd.cut(
    df['YearsCodePro'],
    bins=[-1, 2, 5, 10, 100],
    labels=['Beginner', 'Intermediate', 'Experienced', 'Expert'])

print(df['ExperienceLevel'].value_counts())
```
